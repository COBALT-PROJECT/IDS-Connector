## A tiny utility function

```
def add(a, b):
    """Return the sum of a and b."""
    return a + b

add(2, 3)
```

## Plot a quick chart (matplotlib only)

```
import matplotlib.pyplot as plt
data = [1, 3, 2, 5, 7, 6]
plt.figure()
plt.plot(data)
plt.title('Quick Line Plot')
plt.xlabel('Index')
plt.ylabel('Value')
plt.show()
```
